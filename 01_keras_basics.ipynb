# Session 1: Neural Network programming

## Overview and objectives

This session is a practical introduction to **neural network programming**. The main goal is to connect the abstract idea of a neural network with the concrete tools used to implement, train, and evaluate modern AI models, and get experience with the standard pipeline.

By the end of the session, you should be able to:

- Understand the basic abstraction behind neural networks: layers, activations, parameters, and predictions.
- Explain why neural networks became so successful, as well as their main strengths and limitations.
- Build and train simple models in **Keras**, using the standard ML pipeline.
- Distinguish between **model complexity** and **data representation** as two different sources of performance.
- Identify common neural architectures for **tabular data**, **images**, and **time series**.

# Neural Network abstraction

- Neural networks are a very **popular family of ML models**.
- Conceptually, they can be seen as a generalized **ensemble of logistic regression** models.

## Connectionist model built from complex interactions of simple units

- **Basic Model**: Multi-layer perceptron (MLP). Dense connections of units across multiple layers:
    - **Neurons**: Basic units that receive input, process it, and pass on the output.
    - **Layers**: 
      - **Input Layer**: Where data enters.
      - **Hidden Layers**: Intermediate layers where computation happens.
      - **Output Layer**: Final layer that outputs results.
    - **Activation Function**: Determines the output of neurons; common functions include Sigmoid, Tanh, and ReLU.

## Advantages:
 - **Hardware-friendly**: Regular structures, well suited for optimization (tensor-based computations, parallelization).
 - **Scalability**: You can train **huge models** on **huge datasets** (trillions of parameters over TBs of data).
 - **They can manage unstructured data**: Usually you don't need **feature engineering**, they can extract them automatically from the raw data (images, audio, text...).

## Disadvantages:
 - **Data-hungry**: Their benefits are usually seen only on **extremely large** datasets.
 - **Black boxes**: Predictions cannot be explained or interpreted. Potential unexpected behaviors. No guarantees.
 - **Poor handling of discontinuities**: Smoothness is assumed in the feature space and output. Difficulties with discrete problems (rule-based classificiations, counting, etc.)

## 3: The relevance of feature engineering: Reading a Clock

📝 Task: **given a representation of an analogue clock, predict the time of day (in hours and minutes)**.

We have a dataset containing three input representations of the *same* information:

1. **Raw pixel images** (64×64 grayscale bitmaps): 🧗 hard.
2. **Cartesian coordinates** of the tips of the two hands: 🏃 much easier.
3. **Polar angles** of the two hands: 🚶 trivial.

The objective is to assess the real impact of the input representation in the learnability of a model, and even in the possibility to address the problem using Machine Learning.

### Dataset Loading and Exploration:

The dataset is already available in the course repository in HDF5 format. Let's load it and explore how it is organized and how data looks like:

```python
import h5py

with h5py.File("./assets/clock_dataset.h5", "r") as f:
    images  = f["images"][...]    # (720, 64, 64)  float32
    coords  = f["coords"][...]    # (720, 4)       float32
    angles  = f["angles"][...]    # (720, 2)       float32
    targets = f["targets"][...]   # (720, 2)       float32
    description = f.attrs["description"]

print(description)
fig, axes = plt.subplots(1, 6, figsize=(12, 4))
sample_times = [0, 90, 375, 555, 665, 719]
for ax, t in zip(axes.flat, sample_times):
    ax.imshow(images[t], cmap="gray")
    h, m = targets[t]
    h = int(h*12.)
    m = int(m*60.)
    ax.set_title(f"{h:02d}:{m:02d}", fontsize=9)
    ax.set_yticks([])
    ax.set_xticks([])
    ax.set_xlabel(f'Coords: [{coords[t][0]:.2f}, {coords[t][1]:.2f}, {coords[t][2]:.2f}, {coords[t][3]:.2f}]\nAngles: [{angles[t][0]:.2f}, {angles[t][1]:.2f}]', fontsize=7)
plt.tight_layout(); plt.show()
```

### Model Training and Evaluation

 - We keep the **same data split** across the three experiments so that the only varying factor is the input representation.
 - From a total of 720 samples, we take 80% for training (576) and 20% for testing (144).
 - ⚠️ **Important:** We want to predict 2 values with different meaning (hours and minutes). Prediction errors on each output should be weighted differently, and therefore we need a **custom loss function**.

```python
N = len(targets)
perm = np.random.permutation(N)
train_idx, test_idx = perm[:576], perm[576:]
y_train, y_test = targets[train_idx], targets[test_idx]

def loss_mae_minutes(y_true, y_pred):
    """Custom loss function, measuring the error in minutes."""
    time = y_true[:,0]*720+y_true[:,1]*60
    pred = y_pred[:,0]*720.+y_pred[:,1]*60.
    return ops.mean(ops.abs(time-pred))
```

#### First model: Complex CNN trained on raw images:

```python
x_train, x_test = images[train_idx, :, :, None], images[test_idx, :, :, None]

cnn = keras.Sequential([
    keras.layers.Input(x_train[0].shape),
    keras.layers.Conv2D(32, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(64, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(128, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(2),  # (hour, minute)
])
cnn.summary()
```

```python
cnn.compile(optimizer="adam", loss=loss_mae_minutes)
hist_cnn = cnn.fit(x_train, y_train,
                   validation_split=0.2,
                   epochs=60, batch_size=32)
pred = cnn.predict(x_test, verbose=0)
print(f"\n[CNN on images] - Test MAE = {loss_mae_minutes(y_test, pred):.1f} minutes")
```

#### Second model: Multi-layer MLP on hand coordinates

```python
x_train, x_test = coords[train_idx], coords[test_idx]

mlp_coords = keras.Sequential([
    keras.layers.Input((4,)),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(2),
])
mlp_coords.summary()
mlp_coords.compile(optimizer="adam", loss=loss_mae_minutes)
hist_coords = mlp_coords.fit(x_train, y_train,
                             validation_split=0.2,
                             epochs=60, batch_size=32, verbose=0)

pred = mlp_coords.predict(x_test, verbose=0)
print(f"\n[MLP on coords]      val MAE = {loss_mae_minutes(y_test, pred):.1f} minutes")
```

#### Third model: Logistic regressor on hand angles.

```python
x_train, x_test = angles[train_idx], angles[test_idx]

mlp_angles = keras.Sequential([
    keras.layers.Input((2,)),
    keras.layers.Dense(8, activation="relu"),
    keras.layers.Dense(2),
])
mlp_angles.summary()
mlp_angles.compile(optimizer="adam", loss=loss_mae_minutes)
hist_angles = mlp_angles.fit(x_train, y_train,
                             validation_split=0.2,
                             epochs=60, batch_size=32, verbose=0)

pred = mlp_angles.predict(x_test, verbose=0)
print(f"\n[MLP on angles]      Test MAE = {loss_mae_minutes(y_test, pred):.1f} minutes")
```

```python
plt.plot(hist_cnn.history['val_loss'], label='CNN')
plt.plot(hist_coords.history['val_loss'], label='Coords')
plt.plot(hist_angles.history['val_loss'], label='Angles')
plt.yscale('log')
plt.ylabel('Error (Minutes)')
plt.xlabel('Epoch')
plt.legend()
plt.title('Evolution of the validation loss for different models (log scale)')
```

## 4: Time Series classification

Time series analysis is pervasive on Artificial Intelligence and Machine Learning applications (energy, finances, medicine,...). It has some distinguishing features:
 - Usually, the time dimension can only be analyzed in **one direction** (future information cannot be exploited).
 - Problems are usually **highly unbalanced**.

### Example dataset: The FordA diagnosis

This data was originally used in a competition in the IEEE World Congress on Computational Intelligence, 2008. The classification problem is to **diagnose** whether a certain symptom exists or does not exist in an **automotive subsystem**. The train and test data were collected in typical operating conditions, with **minimal noise contamination**. 
 - The dataset contains 3601 training instances and another 1320 testing instances.
 - Each sample consists of 500 measurements of engine noise and a binary classification.

The data is already prepared for simple loading and processing:

```python
def readFordA(filename):
    """Utility function for loading the FordA dataset"""
    data = np.loadtxt(filename, delimiter="\t")
    y = data[:, 0]
    x = data[:, 1:]
    return x, y.astype(int)


root_url = "/opt/tljh/common/FordA_dataset/"

x_train, y_train = readFordA(root_url + "FordA_TRAIN.tsv")
x_test, y_test = readFordA(root_url + "FordA_TEST.tsv")

x_train = x_train.reshape((x_train.shape[0], x_train.shape[1], 1))
x_test = x_test.reshape((x_test.shape[0], x_test.shape[1], 1))

n_classes = len(np.unique(y_train))

#Shuffle the training set
idx = np.random.permutation(len(x_train))
x_train = x_train[idx]
y_train = y_train[idx]

#Convert the -1 labels to 0
y_train[y_train == -1] = 0
y_test[y_test == -1] = 0
```

### NNs for Time Series in theory: Recurrent Neural Networks

The most mathematically sound model for dealing with temporal data are Recurrent Neural Networks (RNNS):
 - They include **backward connections** in the architecture to keep information from past observations.
 - They are **sequential** in nature. Past observations condition future predictions.
 - Very **difficult to train**. Specialized architectures such as [LSTM](https://www.mathworks.com/discovery/lstm.html) or [GRUs](https://en.wikipedia.org/wiki/Gated_recurrent_unit) try to mitigate practical issues.

In practice, building a RNN model in Keras is equivalent to any other architecture. For this problem, we create a basic network with the following structure:
 - A **Convolutional front-end** that analyzes the raw time series and performs downsampling while providing short-term temporal patterns (5-10 samples).
 - Two **RNN layers** (`GRUs`) that analyze the temporal data at different scales, capturing possible long-term relations and providing a suitable representation for direct classification.

```python
rnn_model = keras.Sequential([
    keras.layers.Input(x_train[0].shape),
    #CNN front-end: Reduces the dimension of the input time series from 500 -> 125, while 
    #learning relevant short-term temporal patterns.
    keras.layers.Conv1D(64, kernel_size=5, strides=2, padding="same", activation="relu"),
    keras.layers.Conv1D(64, kernel_size=5, strides=2, padding="same", activation="relu"),
    #RNN Layers (GRU with CUDNN settings for highest performance)
    keras.layers.GRU(64, return_sequences=True,
               activation="tanh",
               recurrent_activation="sigmoid",
               reset_after=True, unroll=False),
    keras.layers.GRU(32, return_sequences=False,
               activation="tanh",
               recurrent_activation="sigmoid",
               reset_after=True, unroll=False),
    #Final classifier
    keras.layers.Dense(1, activation="sigmoid")
])
rnn_model.summary()
```

Let's evaluate this architecture with a standard pipeline:

```python
rnn_model.compile(
    optimizer=keras.optimizers.Adam(),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

rnn_history = rnn_model.fit(
    x_train, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=64
)

test_loss, test_acc = rnn_model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")
print(f"Test loss:     {test_loss:.4f}")
```

### NNs for Time Series in practice: Transformers

Sequential training has been a major drawback for scaling RNN models. That's the main reason for the appearance in 2017 of the **most successful architecture** of the last decade: The Transformer. Their basic advantages are:
 - **Parallel Processing:** Transformers efficiently process entire sequences at once, leading to faster computations and better performance with large datasets.
 - **Long-Range Dependencies:** Self-attention captures relationships between distant parts of a sequence more effectively than RNNs, which struggle with long-range dependencies.
 - **Scalability and Performance:** Transformers achieve state-of-the-art results in various tasks and scale well with increased data and model size, surpassing the capabilities of traditional RNNs.

This, together with the spectacular surge of Large Language Models (LLMs) for natural language processing applications (and the hype...), has led to a quick replacement of RNNs by Transformers in most domains.

#### Building the model

We will implement the basic transfomer architecture as described in the foundational article ["Attention is all you need"](https://arxiv.org/abs/1706.03762).

<img src="https://upload.wikimedia.org/wikipedia/commons/3/34/Transformer%2C_full_architecture.png" style="margin:auto" width="60%"/>

Our model processes a tensor of shape `(batch size, sequence length, features)`,
where `sequence length` is the number of time steps and `features` is each input
timeseries.

We include residual connections, layer normalization, and dropout.
The resulting layer can be stacked multiple times.

```python
# ------------------------------------------------------------------
# Transformer block
# ------------------------------------------------------------------
def transformer_block(x, d_model, num_heads, ff_dim, dropout=0.1):
    # --- Multi-headed self-attention ---
    h = keras.layers.LayerNormalization(epsilon=1e-6)(x)
    h = keras.layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
            dropout=dropout,
        )(h, h)
    x = keras.layers.Add()([x, h])

    # --- Feed-forward network ---
    h = keras.layers.LayerNormalization(epsilon=1e-6)(x)
    h = keras.layers.Dense(ff_dim, activation="gelu")(h)
    h = keras.layers.Dropout(dropout)(h)
    h = keras.layers.Dense(d_model)(h)
    x = keras.layers.Add()([x, h])
    return x

# ------------------------------------------------------------------
# Full model
# ------------------------------------------------------------------
def build_transformer_model(input_shape, d_model=48, num_heads=4, ff_dim=128,
                num_blocks=2, patch_size=8, dropout=0.1):
    inputs = keras.Input(shape=input_shape)

    # Same 1D convolutional front-end than for the RNN. Downsamples sequence length 500 -> 125
    x = keras.layers.Conv1D(d_model, kernel_size=5, strides=2, padding="same",
                      activation="relu")(inputs)
    x = keras.layers.Conv1D(d_model, kernel_size=5, strides=2, padding="same",
                      activation="relu")(x)

    # --- Learned positional embedding ---
    seq_len = x.shape[1]
    positions = ops.arange(0, seq_len)
    pos_emb = keras.layers.Embedding(seq_len, d_model)(positions)
    x = x + pos_emb

    x = keras.layers.Dropout(dropout)(x)

    # --- Stacked Transformer blocks ---
    for _ in range(num_blocks):
        x = transformer_block(x, d_model, num_heads, ff_dim, dropout)

    x = keras.layers.LayerNormalization(epsilon=1e-6)(x)
    x = keras.layers.GlobalAveragePooling1D()(x)
    x = keras.layers.Dropout(dropout)(x)
    outputs = keras.layers.Dense(1, activation="sigmoid")(x)

    return keras.Model(inputs, outputs)

tr_model = build_transformer_model(x_train.shape[1:])
tr_model.summary()
```

Now we compare the performance of the Transformer model with the reference RNN implementation in the same setup:

```python
tr_model.compile(
    optimizer=keras.optimizers.Adam(),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

tr_history = tr_model.fit(
    x_train, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=64
)

test_loss, test_acc = tr_model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")
print(f"Test loss:     {test_loss:.4f}")
```

### Looking from another angle: The Frequency Domain

From the initial exploration of the dataset, we couldn't observe any meaningful difference in the time distribution of the target classes:

<img src="./assets/fordA_ts_distribution.png" width="50%" style="margin:auto"/>

However, what happens if we look at the data from a different "perspective"? Let's consider the frequency content of each time series by means of the Fast Fourier Transform (FFT):

```python
def to_fft(x):
    """Converts a time series to the frequency domain by means of the FFT. Focuses on magnitude"""
    X = np.fft.rfft(x.squeeze(), axis=1)     # rfft → half-spectrum
    X = np.abs(X)                            # only interested in magnitude
    # Log-compress: spectral energy spans many orders of magnitude
    X = np.log1p(X)
    return X[..., np.newaxis].astype("float32")

x_train_fft = to_fft(x_train)
x_test_fft  = to_fft(x_test)

pos = x_train_fft[y_train==1]
neg = x_train_fft[y_train==0]
tpos = np.tile(np.arange(pos.shape[1]), pos.shape[0])
tneg = np.tile(np.arange(neg.shape[1]), neg.shape[0])
sns.lineplot(x=tpos, y=pos.flatten(), errorbar='sd', label='Class 1')
sns.lineplot(x=tneg, y=neg.flatten(), errorbar='sd', label='Class 0')
```

From this perspective, the difference between classes is evident. We can even see that all the discriminative information is in the first ~75 components, allowing us to further compress the representation and simplify the model:

```python
#Cut the FFT to keep only the first 75 components.
x_train_fft = x_train_fft[:,:75,:]
x_test_fft = x_test_fft[:,:75,:]
#Re-plot the distribution.
pos = x_train_fft[y_train==1]
neg = x_train_fft[y_train==0]
tpos = np.tile(np.arange(pos.shape[1]), pos.shape[0])
tneg = np.tile(np.arange(neg.shape[1]), neg.shape[0])
sns.lineplot(x=tpos, y=pos.flatten(), errorbar='sd', label='Class 1')
sns.lineplot(x=tneg, y=neg.flatten(), errorbar='sd', label='Class 0')
```
